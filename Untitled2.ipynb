```python

from IPython.display import HTML, display

game = r"""
<div id="game" style="max-width:800px;margin:auto;background:#10152b;
padding:15px;border-radius:15px;color:white;text-align:center;
font-family:Arial;">

<h2 style="color:#00ffcc;">⚡ NEON RUNNER ⚡</h2>
<p>Jump over obstacles and collect coins!</p>

<canvas id="canvas" width="760" height="350"
style="width:100%;background:#090d20;border:2px solid #00ffcc;
border-radius:10px;max-width:760px;"></canvas>

<h3 id="score">Score: 0 | Coins: 0</h3>

<button onclick="jump()" style="padding:12px 35px;
background:#00ffcc;border:0;border-radius:8px;
font-weight:bold;cursor:pointer;">JUMP ⬆</button>

<button onclick="restart()" style="padding:12px 25px;
background:#ff4d9d;border:0;border-radius:8px;
font-weight:bold;cursor:pointer;color:white;">RESTART ↻</button>

<p style="color:#aaa;">Keyboard: SPACE / ↑ to Jump</p>
</div>

<script>
const canvas = document.getElementById("canvas");
const ctx = canvas.getContext("2d");

let player, obstacles, coins, score, collected;
let speed, frame, over;

function restart(){
    player = {x:90,y:255,w:38,h:50,vy:0,ground:true};
    obstacles = [];
    coins = [];
    score = 0;
    collected = 0;
    speed = 5;
    frame = 0;
    over = false;
    loop();
}

function jump(){
    if(over){
        restart();
        return;
    }
    if(player.ground){
        player.vy = -13;
        player.ground = false;
    }
}

document.addEventListener("keydown",e=>{
    if(e.code==="Space" || e.code==="ArrowUp"){
        e.preventDefault();
        jump();
    }
});

canvas.addEventListener("touchstart",e=>{
    e.preventDefault();
    jump();
},{passive:false});

canvas.addEventListener("click",jump);

function drawPlayer(){
    ctx.fillStyle="#00ffcc";
    ctx.shadowColor="#00ffcc";
    ctx.shadowBlur=15;
    ctx.fillRect(player.x,player.y,player.w,player.h);
    ctx.shadowBlur=0;

    ctx.fillStyle="#10152b";
    ctx.fillRect(player.x+23,player.y+10,7,7);

    ctx.fillStyle="#ffffff";
    ctx.fillRect(player.x+25,player.y+12,3,3);
}

function loop(){
    ctx.clearRect(0,0,760,350);

    // Background
    ctx.fillStyle="#111b38";
    for(let i=0;i<8;i++){
        let x=i*110;
        ctx.fillRect(x,150,65,130);
        ctx.fillStyle="#24365b";
        for(let j=0;j<3;j++){
            ctx.fillRect(x+12+j*17,170,7,12);
        }
        ctx.fillStyle="#111b38";
    }

    // Ground
    ctx.fillStyle="#00ffcc";
    ctx.fillRect(0,305,760,3);

    if(!over){
        frame++;
        score++;
        speed=5+Math.floor(score/500);

        // Gravity
        player.vy+=0.65;
        player.y+=player.vy;

        if(player.y+player.h>=305){
            player.y=305-player.h;
            player.vy=0;
            player.ground=true;
        }

        // Obstacles
        if(frame%85===0){
            obstacles.push({x:760,y:275,w:30,h:30});
        }

        // Coins
        if(frame%110===0){
            coins.push({x:760,y:210,w:18,h:18});
        }

        obstacles.forEach(o=>o.x-=speed);
        coins.forEach(c=>c.x-=speed);

        obstacles=obstacles.filter(o=>o.x+o.w>0);
        coins=coins.filter(c=>c.x+c.w>0);

        // Collision
        obstacles.forEach(o=>{
            if(player.x<o.x+o.w &&
               player.x+player.w>o.x &&
               player.y<o.y+o.h &&
               player.y+player.h>o.y){
                over=true;
            }
        });

        coins=coins.filter(c=>{
            if(player.x<c.x+c.w &&
               player.x+player.w>c.x &&
               player.y<c.y+c.h &&
               player.y+player.h>c.y){
                collected++;
                score+=50;
                return false;
            }
            return true;
        });
    }

    // Draw player
    drawPlayer();

    // Draw obstacles
    ctx.fillStyle="#ff4081";
    ctx.shadowColor="#ff4081";
    ctx.shadowBlur=12;
    obstacles.forEach(o=>ctx.fillRect(o.x,o.y,o.w,o.h));
    ctx.shadowBlur=0;

    // Draw coins
    ctx.fillStyle="#ffe600";
    ctx.shadowColor="#ffe600";
    ctx.shadowBlur=12;
    coins.forEach(c=>{
        ctx.beginPath();
        ctx.arc(c.x+9,c.y+9,9,0,Math.PI*2);
        ctx.fill();
    });
    ctx.shadowBlur=0;

    document.getElementById("score").innerText=
        "Score: "+score+" | Coins: "+collected;

    if(over){
        ctx.fillStyle="rgba(0,0,0,0.75)";
        ctx.fillRect(0,0,760,350);
        ctx.textAlign="center";
        ctx.fillStyle="#ff4081";
        ctx.font="bold 42px Arial";
        ctx.fillText("GAME OVER!",380,150);
        ctx.fillStyle="white";
        ctx.font="22px Arial";
        ctx.fillText("Final Score: "+score,380,195);
        ctx.fillText("Press RESTART to play again",380,240);
        return;
    }

    requestAnimationFrame(loop);
}

restart();
</script>
"""

display(HTML(game))
```
